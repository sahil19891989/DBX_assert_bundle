```python
from pyspark.sql import functions as F

bronze = (
    spark.read.option("header", True).option("inferSchema", True)
    .csv("/Volumes/metlife_dmo/lapse/raw_files/policies.csv")
    .withColumn("_ingest_ts", F.current_timestamp())          # when we loaded it
    .withColumn("_source_file", F.col("_metadata.file_path"))  # where it came from
)
bronze.write.mode("overwrite").saveAsTable("metlife_dmo.lapse.bronze_policies")
```
